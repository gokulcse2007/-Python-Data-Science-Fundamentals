1. VARIABLES AND DATA TYPES

Variables are used to store data in Python.

Python supports different data types such as:
1. Integer
2. Float
3. String
4. Boolean

```python
# Variables in Python

name = "Gokul"
age = 18
department = "CSE"
college = "Chennai Institute of Technology"

print("Name:", name)
print("Age:", age)
print("Department:", department)
print("College:", college)
```

```python
# Different Data Types in Python

a = 10
b = 10.5
c = "Python"
d = True

print("Integer:", type(a))
print("Float:", type(b))
print("String:", type(c))
print("Boolean:", type(d))
```

Operators are special symbols used to perform operations on variables and values.

Types of operators:
1. Arithmetic Operators
2. Comparison Operators
3. Logical Operators
4. Assignment Operators

```python
# Arithmetic Operators

a = 20
b = 10

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
```

```python
# Comparison Operators

a = 20
b = 10

print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
print("Greater or Equal:", a >= b)
print("Less or Equal:", a <= b)
```

```python
# Assignment Operators

a = 10

a += 5
print("Addition Assignment:", a)

a -= 3
print("Subtraction Assignment:", a)

a *= 2
print("Multiplication Assignment:", a)
```

3. CONDITIONAL STATEMENTS

Conditional statements are used to make decisions
based on conditions.

Types:
1. If statement
2. If-else statement
3. If-elif-else statement

```python
# If-Else Statement

marks = 85

if marks >= 50:
    print("Student is Pass")
else:
    print("Student is Fail")
```

Loops are used to execute a block of code repeatedly.

Types of loops:
1. For Loop
2. While Loop

Loops reduce code repetition and make programming easier.

```python
# For Loop

for i in range(1, 11):
    print(i)
```

```python
# While Loop

i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# Sum of First 10 Natural Numbers

total = 0

for i in range(1, 11):
    total = total + i

print("Sum:", total)
```

5. FUNCTIONS IN PYTHON

A function is a block of reusable code that performs a
specific task.

Advantages:
1. Code reusability.
2. Reduces code repetition.
3. Makes programs easier to understand.

Functions are created using the def keyword.

```python
# Function to Add Two Numbers

def addition(a, b):
    return a + b

result = addition(10, 20)

print("Addition:", result)
```

```python
# Function to Find Square

def square(n):
    return n * n

result = square(5)

print("Square:", result)
```

```python
# Even or Odd Using Function

def check_number(n):
    if n % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")

check_number(10)
check_number(7)
```

Lists:
A list is used to store multiple values in a single variable.
Lists are ordered and changeable.

Dictionaries:
A dictionary stores data in key-value pairs.
Each value is accessed using its key.

```python
# Lists in Python

students = ["Arun", "Kumar", "Ravi", "Ram"]

print("Student List:", students)

print("First Student:", students[0])

print("Second Student:", students[1])

print("Total Students:", len(students))
```

```python
# List Operations

marks = [85, 90, 75, 80]

marks.append(95)

print("After Adding:", marks)

marks.remove(75)

print("After Removing:", marks)

print("Total:", sum(marks))

print("Average:", sum(marks) / len(marks))
```

```python
# Dictionaries in Python

student = {
    "Name": "Gokul",
    "Age": 18,
    "Department": "CSE",
    "Marks": 90
}

print("Student Details:", student)

print("Name:", student["Name"])

print("Age:", student["Age"])

print("Department:", student["Department"])

print("Marks:", student["Marks"])
```

```python
# Dictionary Operations

student = {
    "Name": "Gokul",
    "Marks": 90
}

# Adding a new value
student["Department"] = "CSE"

# Updating marks
student["Marks"] = 95

print("Updated Student Details:")

print(student)
```

## 7. FILE HANDLING IN PYTHON

File handling allows Python to create, read, write and modify files.

Types of file operations:

1. Read (r) – Reads data from a file.
2. Write (w) – Writes data into a file.
3. Append (a) – Adds new data to an existing file.

```python
# Creating and Writing a File

with open("student.txt", "w") as file:
    file.write("Name: Gokul\n")
    file.write("Department: CSE\n")
    file.write("Marks: 90\n")

print("File created successfully!")
```

```python
# Reading a File

with open("student.txt", "r") as file:
    content = file.read()

print("Student Details:")
print(content)
```

```python
# Adding Data to an Existing File

with open("student.txt", "a") as file:
    file.write("College: Chennai Institute of Technology\n")

print("Data added successfully!")
```

```python
# Reading Updated File

with open("student.txt", "r") as file:
    content = file.read()

print(content)
```

10. INTRODUCTION TO PANDAS

Pandas is a Python library used for data analysis.

Features of Pandas:
1. Creates tables using DataFrames.
2. Organises data into rows and columns.
3. Helps analyse large datasets.
4. Handles missing data.

```python
# Introduction to Pandas

import pandas as pd

data = {
    "Name": ["Arun", "Kumar", "Ravi", "Ram"],
    "Age": [18, 19, 18, 20],
    "Marks": [85, 90, 75, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Data Analysis Using Pandas

print("Total Marks:", df["Marks"].sum())

print("Average Marks:", df["Marks"].mean())

print("Highest Marks:", df["Marks"].max())

print("Lowest Marks:", df["Marks"].min())
```

11. INTRODUCTION TO MATPLOTLIB

Matplotlib is a Python library used for data visualization.

Types of charts:
1. Bar Chart
2. Line Chart
3. Pie Chart
4. Histogram

It helps represent data visually.

```python
# Bar Chart Using Matplotlib

import matplotlib.pyplot as plt

names = ["Arun", "Kumar", "Ravi", "Ram"]

marks = [85, 90, 75, 95]

plt.bar(names, marks)

plt.xlabel("Students")

plt.ylabel("Marks")

plt.title("Student Marks Analysis")

plt.show()
```

```python
# Pie Chart Using Matplotlib

subjects = ["Python", "Maths", "Science"]

marks = [85, 90, 75]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")

plt.title("Subject Marks Distribution")

plt.show()
```

STUDENT PERFORMANCE ANALYSIS USING PYTHON

Objective:

Project Features:

1. IMPORTING REQUIRED LIBRARIES:
Libraries are collections of predefined functions
that help us perform different operations.

NumPy – Numerical calculations.
Pandas – Data analysis.
Matplotlib – Data visualization.

```python
# Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All Libraries Imported Successfully!")
```

2. CREATING STUDENT DATASET:
A dataset is a collection of related information.

In this project, we create a dataset containing
student names and their marks in three subjects.

Subjects:
1. Python
2. Maths
3. Science

```python
# Creating Student Dataset

data = {
    "Name": ["Arun", "Kumar", "Ravi", "Ram", "Ajay"],
    "Python": [85, 90, 75, 95, 80],
    "Maths": [80, 85, 70, 90, 75],
    "Science": [90, 95, 80, 85, 85]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)
```

3. CALCULATING TOTAL AND AVERAGE MARKS

Explanation:
In this step, we calculate the total marks and
average marks of every student.

Total = Python + Maths + Science

Average = Total / 3

```python
# Calculate Total and Average Marks

df["Total"] = df["Python"] + df["Maths"] + df["Science"]

df["Average"] = df["Total"] / 3

print(df)
```

4. IDENTIFYING THE CLASS TOPPER

Explanation:
In this step, we identify the student who scored
the highest total marks in the class.

```python
# Finding Class Topper

highest_marks = df["Total"].max()

topper = df[df["Total"] == highest_marks]

print("Class Topper Details:")
print(topper[["Name", "Total", "Average"]])
```

5. STUDENT PERFORMANCE ANALYSIS

Explanation:
In this step, we analyse the overall student performance.

We calculate:
1. Class Average
2. Highest Marks
3. Lowest Marks
4. Total Number of Students

```python
# Student Performance Analysis

class_average = np.mean(df["Average"])

highest = df["Total"].max()

lowest = df["Total"].min()

total_students = len(df)

print("Class Average:", round(class_average, 2))

print("Highest Marks:", highest)

print("Lowest Marks:", lowest)

print("Total Students:", total_students)
```

6. DATA VISUALIZATION USING BAR CHART

Explanation:
Data visualization is the process of representing
data using graphs and charts.

A bar chart is used to compare the average marks
of different students.

```python
# Student Performance Bar Chart

plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Average"])

plt.xlabel("Student Name")

plt.ylabel("Average Marks")

plt.title("Student Performance Analysis")

plt.show()
```

7. EXPORTING DATASET TO CSV FILE

Explanation:
CSV stands for Comma Separated Values.

It is used to store tabular data in a file format
that can be opened using Excel and other applications.

```python
# Saving Dataset as CSV

df.to_csv("student_performance.csv", index=False)

print("Dataset saved successfully!")
```
